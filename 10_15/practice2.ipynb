```python
# --- 設定 Matplotlib / Seaborn 中文字型支援（Mac / Windows / Colab 通用）---
import glob
import os
import mplfonts
from matplotlib import font_manager
import matplotlib.pyplot as plt
import seaborn as sns
from mplfonts import use_font

for f in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(f)
use_font("Noto Sans CJK SC")
plt.rcParams['axes.unicode_minus'] = False
print("繁體中文字型已成功載入！")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# 讀取顧客消費資料集
df = pd.read_csv('顧客消費資料集.csv')
print("=== 原始顧客資料 ===")
display(df.head())
```

```python
# 1. 數值特徵縮放
num_cols = ['年齡', '年消費額_元']

# 標準化 (Z-Score)
std_scaler = StandardScaler()
df_std = pd.DataFrame(std_scaler.fit_transform(df[num_cols]), columns=['年齡_標準化', '年消費額_標準化'])

# 歸一化 (0~1)
minmax_scaler = MinMaxScaler()
df_minmax = pd.DataFrame(minmax_scaler.fit_transform(df[num_cols]), columns=['年齡_歸一化', '年消費額_歸一化'])

print("=== 標準化後數據 (平均約0, 標準差約1) ===")
print(df_std.head().round(2))

print("\n=== 歸一化後數據 (壓縮至 0 ~ 1) ===")
print(df_minmax.head().round(2))
```

```python
# 2. 編碼類別特徵
等級順序 = {'銅卡': 1, '白金': 2, '鑽石': 3}

df_encoded = df.copy()
df_encoded['會員等級'] = df_encoded['會員等級'].map(等級順序)
df_encoded = pd.get_dummies(df_encoded, columns=['所在縣市'], prefix=['縣市'])

print("=== 編碼後的新欄位 ===")
print(df_encoded.columns.tolist())
display(df_encoded.head(10))
```
