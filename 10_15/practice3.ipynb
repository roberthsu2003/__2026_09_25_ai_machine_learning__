```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, Ridge

# 讀取廣告與銷售資料集
df = pd.read_csv('廣告與銷售資料集.csv')
X = df[['廣告預算_萬元']]
y = df['產品銷售額_萬元']

print("=== 資料集預覽 ===")
print(df.head())
```

```python
# 1. 訓練普通線性回歸，取得內部參數
lr = LinearRegression()
lr.fit(X, y)

w = lr.coef_[0]
b = lr.intercept_

print("=== 模型自主學出的內部參數 ===")
print(f"學到的權重斜率 w (廣告效益係數): {w:.4f}")
print(f"學到的偏置截距 b (基本保底銷售): {b:.4f}")
print(f"預測方程式: 銷售額 = {w:.2f} * 廣告預算 + {b:.2f}")
```

```python
# 2. 調校外部超參數 (Ridge 正則化懲罰 alpha)
ridge_loose = Ridge(alpha=0.1).fit(X, y)      # 寬鬆超參數
ridge_strict = Ridge(alpha=100.0).fit(X, y)   # 嚴格超參數

print("=== 外部超參數對內部權重的影響 ===")
print(f"寬鬆懲罰 (alpha=0.1) 學出的斜率 w:   {ridge_loose.coef_[0]:.4f}")
print(f"強力懲罰 (alpha=500.0) 學出的斜率 w: {ridge_strict.coef_[0]:.4f} (權重被大幅抑制防止過擬合)")

# 3. 繪圖比較
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='廣告預算_萬元', y='產品銷售額_萬元', color='royalblue', label='真實銷售樣本', s=60)
x_line = pd.DataFrame({X.columns[0]: np.linspace(df['廣告預算_萬元'].min(), df['廣告預算_萬元'].max(), 100)})

plt.plot(x_line[X.columns[0]], lr.predict(x_line), color='green', linewidth=2, label=f'普通線性回歸 (w={w:.2f})')
plt.plot(x_line[X.columns[0]], ridge_strict.predict(x_line), color='crimson', linestyle='--', linewidth=2, label=f'嚴格正則化 Ridge (alpha=500, w={ridge_strict.coef_[0]:.2f})')

plt.title('內部學出參數與外部超參數調節效應', fontsize=14, pad=12)
plt.xlabel('廣告預算 (萬元)', fontsize=12)
plt.ylabel('產品銷售額 (萬元)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("💡 觀念總結：內部參數由數據驅動自主學習，外部超參數則是工程師控制模型行為的控制旋鈕！")
```
